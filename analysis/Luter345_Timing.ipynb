# Calculation of Timing for Luter 345 Cosmic Ray Detector Setup

The Luter 345 Cosmic Ray Detector is a scintillation detector that is used to detect cosmic rays. The detector comprises two parallel scintillator paddles, each with two PMTs - one at each end of light guides that collect scintillation light from the left
 and right end of the scintillating material.
 
Consider a cosmic ray that strikes the top scintillator paddle at position, $x_{top}$ at t=0. The cosmic ray then passes through the top scintillator paddle and strikes the bottom scintillator paddle at position, $x_{bot}$ at time, $t_{bot}$. The times of the signals generated by each PMT are given by:

$$t_{top/right} = \frac{\frac{d}{2} - x_{top}}{v_{scint}} + t_{tube/topright$$

$$t_{top/left} = \frac{\frac{d}{2} + x_{top}}{v_{scint}} + t_{tube/topleft$$

$$t_{bot/right} = TOF + \frac{\frac{d}{2} - x_{bot}}{v_{scint}} + t_{tube/botright$$

$$t_{bot/left} = TOF + \frac{\frac{d}{2} + x_{bot}}{v_{scint}} + t_{tube/botleft$$

where $d$ is the width of the scintillator paddle, $v_{scint}$ is the speed of light in the scintillator, $t_{tube/topright}$, $t_{tube/topleft}$, $t_{tube/botright}$, and $t_{tube/botleft}$ are the transit times of the scintillation light in the light guides, and $TOF$ is the time of flight of the cosmic ray between the top and bottom scintillator paddles.

We assume that the trigger time for a given event carries the timing of the bottom left signal, i.e. $$t_{trigger} = t_{bot/left}$ +t_{trigger-delay}$$.

The TDC unit in the experiment is operated in common stop mode, where the trigger time is the common stop signal for all four PMTs. The TDC times recorded for each PMT are given by:

$$TDC_{top/right} = t_{trigger} - t_{top/right} $$
$$TDC_{top/left} = t_{trigger} - t_{top/left} $$
$$TDC_{bot/right} = t_{trigger} - t_{bot/right} $$
$$TDC_{bot/left} = t_{trigger} - t_{bot/left} $$

Substituting the expressions for $t_{trigger}$, $t_{top/right}$, $t_{top/left}$, $t_{bot/right}$, and $t_{bot/left}$ into the expressions for $TDC_{top/right}$, $TDC_{top/left}$, $TDC_{bot/right}$, and $TDC_{bot/left}$, we get:

$$\vspace{1cm}$$

$$TDC_{top/right} = t_{bot/left} - t_{top/right} + t_{trigger-delay}$$
$$ = TOF + \frac{\frac{d}{2} - x_{bot}}{v_{scint}} + t_{tube/botleft} - \frac{\frac{d}{2} - x_{top}}{v_{scint}} - t_{tube/topright} + t_{trigger-delay}$$
$$ = TOF + \frac{x_{top} - x_{bot}}{v_{scint}} + t_{tube/botleft} - t_{tube/topright} + t_{trigger-delay}$$

$$\vspace{1cm}$$

$$TDC_{top/left} = t_{bot/left} - t_{top/left} + t_{trigger-delay}$$
$$ = TOF + \frac{\frac{d}{2} - x_{bot}}{v_{scint}} + t_{tube/botleft} - \frac{\frac{d}{2} + x_{top}}{v_{scint}} - t_{tube/topleft} + t_{trigger-delay}$$
$$ = TOF + \frac{-x_{top} - x_{bot}}{v_{scint}} + t_{tube/botleft} - t_{tube/topleft} + t_{trigger-delay}$$

$$\vspace{1cm}$$

$$TDC_{bot/right} = t_{bot/left} - t_{bot/right} + t_{trigger-delay}$$
$$ = TOF +  \frac{\frac{d}{2} - x_{bot}}{v_{scint}} + t_{tube/botleft} - TOF - \frac{\frac{d}{2} + x_{bot}}{v_{scint}} - 
t_{tube/botright} + t_{trigger-delay}$$
$$ = \frac{-2 \cdot x_{bot}}{v_{scint}} + t_{tube/botleft} - t_{tube/botright} + t_{trigger-delay}$$

$$\vspace{1cm}$$

$$TDC_{bot/left} = t_{bot/left} - t_{bot/left} + t_{trigger-delay}$$
$$ = t_{trigger-delay}$$

$$\vspace{1cm}$$

If we now look at the time difference between left and right TDC times for the top and bottom scintillator paddles, we get:

$$TDC_{top/right} - TDC_{top/left} = \frac{2 \cdot x_{top}}{v_{scint}} + t_{tube/topleft} - t_{tube/topright}$$
$$TDC_{bot/right} - TDC_{bot/left} = \frac{2 \cdot x_{bot}}{v_{scint}} + t_{tube/botleft} - t_{tube/botright}$$

from which we can calculate the top and bottom positions of the cosmic ray as:

$$x_{top} = \frac{v_{scint}}{2} \cdot (TDC_{top/right} - TDC_{top/left}) + \frac{t_{tube/topleft} - t_{tube/topright}}{2}$$
$$x_{bot} = \frac{v_{scint}}{2} \cdot (TDC_{bot/right} - TDC_{bot/left}) + \frac{t_{tube/botleft} - t_{tube/botright}}{2}$$

Looking at the time sums for the top and bottom scintillator paddles, we get:

$$TDC_{top/right} + TDC_{top/left} = 2 \cdot TOF - 2 \frac{\cdot x_{bot}}{v_{scint}} +2 t_{tube/botleft} - t_{tube/topleft} - t_{tube/topright} + 2 \cdot t_{trigger-delay}$$
$$TDC_{bot/right} + TDC_{bot/left} = - 2 \frac{\cdot x_{bot}}{v_{scint}} + t_{tube/botleft} + t_{tube/botright} + 2 \cdot t_{trigger-delay}$$

Finally, we can calculate the overall mean time of flight of the cosmic ray as:

$$Mean TOF = \frac{TDC_{top/right} + TDC_{top/left}}{2} - \frac{TDC_{bot/right} + TDC_{bot/left}}{2}$$
$$ = TOF + \frac{t_{tube/botleft} + t_{tube/botright} - t_{tube/topleft} - t_{tube/topright}}{2}$$

If the tube times are the same for all tubes, this should just be equal to the time-of-flight (which may or may not be the case).  But, if we look at this quantity as a function of the angle of the cosmic ray, we should see a correlation between the angle and the time of flight.
